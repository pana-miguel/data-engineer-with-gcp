### 🔐 Configuración de credenciales de Google Cloud

Antes de poder utilizar funciones que interactúan con servicios de **Google Cloud Platform (GCP)**, es necesario configurar las credenciales de autenticación mediante una **cuenta de servicio**.

Asegúrate de contar con el archivo de credenciales en formato **`.json`** y ejecuta el siguiente comando en la terminal:

```bash
export GOOGLE_APPLICATION_CREDENTIALS="/ruta/completa/donde/se/encuentra/la/llave.json"
```

````python
from google.cloud import storage

def upload_to_bucket(bucket_name, source_file, destination_obj):
    client = storage.Client.from_service_account_json("llave1.json")
    bucket = client.bucket(bucket_name)
    obj = bucket.blob(destination_obj)
    obj.upload_from_filename(source_file)
    print(f"el archivo {source_file} se ha subido como {destination_obj} al bucket {bucket_name}")
````

````python
def download_from_bucket(bucket_name, source_obj, destination_file):
    client = storage.Client.from_service_account_json("llave1.json")
    bucket = client.bucket(bucket_name)
    obj = bucket.blob(source_obj)
    obj.download_to_filename(destination_file)
    print(f"el archivo {source_obj} del bucket {bucket_name} se ha descargado como {destination_file}")
````

````python
upload_to_bucket("nuevo-bucket1234321", "perro.jpg", "perro.jpg")
````

````python
download_from_bucket("nuevo-bucket1234321", "perro.jpg", "perro_descargado.jpg")
````
